Pattern Printing in python

```python
n = 5
for i in range(n):
    for j in range(n):
        print("*" , end = "" )
    print()
```

```python

n=5
for i in range(1,n+1):
    print("*" * i)
```

```python
n=5
for i in range(n,0,-1):
    print("*" * i)
```

```python
n=5
for i in range(1,n+1):
    for j in range(1,i+1): #this loop will only range till the i limit 
        print(j,end=" ")
    print()
```

```python
n=5
for i in range(1,n+1):
    for j in range(1,i+1): #this loop will only range till the j limit which remains same in the loop
        print(i,end=" ")
    print()
        
```

6️⃣ Pyramid
    *
   ***
  *****
 *******
*********

```python
n = 5

for i in range(n):
    
    for j in range(n - i - 1):
        print(" ", end="")

    
    for j in range(2*i + 1):
        print("*", end="")

    print()
```
